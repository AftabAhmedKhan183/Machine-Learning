### Feature Selection using Filter Methods

Feature selection is a crucial step in machine learning that aims to reduce the number of input variables to the most relevant ones. This helps in reducing overfitting, improving accuracy, and reducing training time.

**Filter methods** are among the simplest and fastest ways to perform feature selection. They rely on statistical measures to score each feature and select or reject them based on these scores. These methods are independent of the machine learning algorithm being used.

Common filter methods include:

1.  **Correlation**: Measures the linear relationship between two variables. Features highly correlated with the target variable (and less correlated with each other) are generally preferred.
2.  **Variance Threshold**: Removes features with low variance, as features with very little variation carry little information.
3.  **Chi-squared test (Chi2)**: Used for selecting categorical features with respect to a categorical target variable. (Not directly applicable here as 'charges' is numerical).
4.  **ANOVA F-value**: Used for selecting numerical features with respect to a categorical target variable.
5.  **Mutual Information**: Measures the dependency between two variables. It can capture non-linear relationships and is applicable to both continuous and discrete features.

Since our target variable `charges` is numerical, we can start by looking at the correlation of each feature with `charges`.

```python
# Calculate the correlation matrix
correlation_matrix = df_cleaned.corr()

# Display correlations with the target variable 'charges'
# Sort them in descending order for better readability
charges_correlation = correlation_matrix['charges'].sort_values(ascending=False)
display(charges_correlation)
```

From the correlation values, we can observe which features have the strongest linear relationship with `charges`. Features with higher absolute correlation values are generally more relevant.

Next, let's consider another filter method: **Variance Thresholding**.

```python
from sklearn.feature_selection import VarianceThreshold

# Separate features (X) and target (y)
X = df_cleaned.drop('charges', axis=1)
y = df_cleaned['charges']

# Initialize VarianceThreshold with a threshold (e.g., 0 for removing features with zero variance)
# You might need to adjust the threshold based on your dataset
selector = VarianceThreshold(threshold=0)

# Fit and transform the features
X_high_variance = selector.fit_transform(X)

# Get the names of the selected features
selected_features_variance = X.columns[selector.get_support()]

print(f"Original number of features: {X.shape[1]}")
print(f"Number of features after Variance Thresholding: {X_high_variance.shape[1]}")
print("Selected features:")
print(selected_features_variance.tolist())
```

FEATURE SELECTION (Last Step)

```python
from scipy.stats import pearsonr

#Pearson Correlation Calculation

cols = ['age', 'is_female', 'bmi', 'children', 'is_smoker',
       'region_northeast', 'region_northwest', 'region_southeast',
       'region_southwest', 'bmi_category_Underweight', 'bmi_category_Normal',
       'bmi_category_Overweight', 'bmi_category_Obese']

correlations = {}
for col in cols:
  corr,_ = pearsonr(df_cleaned[col],df_cleaned['charges'])
  correlations[col] = corr

correlations_df = pd.DataFrame(list(correlations.items()),columns=['Feature','Correlation'])
correlations_df = correlations_df.sort_values(by='Correlation',ascending=False)
display(correlations_df)
```

```python
df_cleaned.columns
```

```python
cat_features = [
    'is_female', 'is_smoker', 'region_northeast', 'region_northwest',
    'region_southeast', 'region_southwest', 'bmi_category_Underweight',
    'bmi_category_Normal', 'bmi_category_Overweight', 'bmi_category_Obese'
]
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05
df_cleaned['charges_bin'] = pd.qcut(df_cleaned['charges'], q=10, labels=False)
chi2_results = {}

for col in cat_features:
    contingency = pd.crosstab(df_cleaned[col], df_cleaned['charges_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```

**MODEL TRAINING AND EVALUATION**


```python
from sklearn.model_selection import train_test_split
```

```python
X = final_df.drop('charges',axis=1)
y = final_df['charges']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test,y_pred)
print("R2 Score:",r2)

n = X_train.shape[0]
p = X_train.shape[1]

adjusted_r2 = 1 - ((1-r2)*(n-1)/(n-p-1))
print("Adjusted R2 Score :",adjusted_r2)
```

```python
print("The R2 Score and Adjusted R2 Score of the model are :")
print(r2)
print(adjusted_r2)

print("\n\n The Model performs very good")
```
