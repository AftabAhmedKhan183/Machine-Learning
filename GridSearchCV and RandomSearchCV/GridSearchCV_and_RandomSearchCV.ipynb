```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
df['species'].value_counts()
```

```python
from sklearn.model_selection import train_test_split

X = df.drop('species',axis=1)
y = df['species']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.33,random_state=42)
```

```python
X_train.head()
```

```python
from sklearn.neighbors import KNeighborsClassifier

model_KNN = KNeighborsClassifier(n_neighbors=3)
```

```python
model_KNN.fit(X_train,y_train)
```

```python
model_KNN.score(X_test,y_test)
```

```python
#By changing the parameters n_neighbors = [3,5,7,9..] the score changes like 0.98,1.0
```

```python
from sklearn.svm import SVC

model_svm = SVC(C=30,kernel='linear',gamma='auto')
model_svm.fit(X_train,y_train)
```

```python
model_svm.score(X_test,y_test)
```

```python
#By changing the parameters kernel = ['rbf','linear'] and C=1,10,20,30 the score changes like 0.98,1.0
```

```python
from sklearn.model_selection import GridSearchCV

classifier1 = GridSearchCV((model_svm),{
    'C' : [1,10,20,30,40],
    'kernel' : ['rbf','linear'],
    'gamma' :['auto','scale']
},cv=5,return_train_score=False)

classifier1.fit(X_train,y_train)
```

```python
classifier1.cv_results_
```

```python
result1 = pd.DataFrame(classifier1.cv_results_)
result1
```

```python
result1[['param_C','param_gamma','param_kernel','mean_test_score']]
```

```python
from sklearn.model_selection import GridSearchCV

classifier2 = GridSearchCV((model_KNN),{
    'n_neighbors' : [3,5,7,9]
},cv=5,return_train_score=False)

classifier2.fit(X_train,y_train)
```

```python
classifier2.cv_results_
```

```python
result2 = pd.DataFrame(classifier2.cv_results_)
result2
```

```python
result2[['param_n_neighbors','mean_test_score']]
```
