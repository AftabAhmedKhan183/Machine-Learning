```python
#Let us create a function for plotting these numerical columns using histplot
def plotting(var,num):
  plt.subplot(2,2,num)
  sns.histplot(df[var],kde=True)

plotting('Age',1)
plotting('RestingBP',2)
plotting('Cholesterol',3)
plotting('MaxHR',4)
plt.tight_layout()
#plt.tight_layout() is used to avoid overlapping of one entity on the other
```

```python
df['Cholesterol'].value_counts()
```

```python
cholesterol_mean = df['Cholesterol'][df['Cholesterol'] != 0].mean()
```

```python
cholesterol_mean
```

```python
df['Cholesterol'] = df['Cholesterol'].replace(0,cholesterol_mean)
df['Cholesterol'] = df['Cholesterol'].astype(int)
df['Cholesterol'].value_counts()
```

```python
df['RestingBP'].value_counts()
```

```python
resting_bp_mean = df['RestingBP'][df['RestingBP'] != 0].mean()

df['RestingBP'] = df['RestingBP'].replace(0,resting_bp_mean)
df['RestingBP'] = df['RestingBP'].astype(int)
df['RestingBP'].value_counts()
```

```python
plotting('Age',1)
plotting('RestingBP',2)
plotting('Cholesterol',3)
plotting('MaxHR',4)

plt.tight_layout()
```
